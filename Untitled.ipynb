```python
import pandas as pd
import numpy as np
from mlxtend.preprocessing import TransactionEncoder

# Sample transactional dataset (list of baskets)
transactions = [
    ['Milk', 'Bread', 'Butter'],
    ['Milk', 'Bread', 'Eggs', 'Butter'],
    ['Bread', 'Butter', 'Jam'],
    ['Milk', 'Eggs'],
    ['Milk', 'Bread', 'Butter', 'Coffee'],
    ['Coffee', 'Sugar', 'Milk'],
    ['Bread', 'Jam'],
    ['Milk', 'Bread', 'Eggs', 'Coffee'],
    ['Butter', 'Jam', 'Bread'],
    ['Milk', 'Bread', 'Butter']
]

# Transform transactions into a one-hot encoded boolean DataFrame
te = TransactionEncoder()
te_array = te.fit(transactions).transform(transactions)
df = pd.DataFrame(te_array, columns=te.columns_)

print(df.head())
```

```python
from mlxtend.frequent_patterns import fpgrowth

# Extract frequent itemsets
frequent_itemsets = fpgrowth(df, min_support=0.3, use_colnames=True)

# Add a length column for easier filtering and sort by support
frequent_itemsets['length'] = frequent_itemsets['itemsets'].apply(len)
frequent_itemsets = frequent_itemsets.sort_values(by='support', ascending=False)

print(frequent_itemsets)
```

```python
from mlxtend.frequent_patterns import association_rules

# Generate rules based on confidence
rules = association_rules(
    frequent_itemsets, 
    metric="confidence", 
    min_threshold=0.7
)

# Select and sort the most relevant columns
cols = ['antecedents', 'consequents', 'support', 'confidence', 'lift']
rules_filtered = rules[cols].sort_values(by='lift', ascending=False)

print(rules_filtered.to_string(index=False))
```
